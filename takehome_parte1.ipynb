```python
#a. Cifrado Cesar

def cifrado_cesar(texto, desplazamiento):
    """Cifra o descifra un texto usando el cifrado César."""
    resultado = ""
    for caracter in texto:
        if caracter.isalpha():
            base = ord('A') if caracter.isupper() else ord('a')
            nuevo = (ord(caracter) - base + desplazamiento) % 26
            resultado += chr(base + nuevo)
        else:
            resultado += caracter
    return resultado


Nombre = "Soy Juan Diego"
Frase = "Este es mi parcial"
Nombre_encriptado = cifrado_cesar(Nombre, 4)
Frase_encriptada = cifrado_cesar(Frase, 4)
print(f"Nombre: {Nombre}")
print(f"Nombre Cifrado: {Nombre_encriptado}")
print(f"Frase: {Frase}")
print(f"Frase Cifrada: {Frase_encriptada}")

Nombre_descifrado = cifrado_cesar(Nombre_encriptado, -4)
print(f"Nombre Descifrado: {Nombre_descifrado}")
Frase_descifrada = cifrado_cesar(Frase_encriptada, -4)
print(f"Frase Descifrada: {Frase_descifrada}")

#Ataque de fuerza bruta
for Combinacion in range(26):
        posible_Nombre = cifrado_cesar(Nombre_encriptado, -Combinacion)
        posible_Frase = cifrado_cesar(Frase_encriptada, -Combinacion)
        print(f"Nombre {Combinacion:2}: {posible_Nombre}")
        print(f"Frase {Combinacion:2}: {posible_Frase}")



```

```python
#Cifrado Vigenère

def cifrado_vigenere(texto, clave, modo='cifrar'):
    resultado = ""
    clave = clave.upper()
    j = 0  # índice para recorrer la clave

    for caracter in texto:
        if caracter.isalpha():
            base = ord('A') if caracter.isupper() else ord('a')
            desplazamiento = ord(clave[j % len(clave)]) - ord('A')

            if modo == 'descifrar':
                desplazamiento = -desplazamiento

            nuevo = (ord(caracter) - base + desplazamiento) % 26
            resultado += chr(base + nuevo)
            j += 1
        else:
            resultado += caracter

    return resultado


mensaje = "Los Pinguinos"
clave = "Puffle"
cifrado = cifrado_vigenere(mensaje, clave, 'cifrar')
print(f"Mensaje original: {mensaje}")
print(f"Clave: {clave}")
print(f"Cifrado: {cifrado}")

descifrado = cifrado_vigenere(cifrado, clave, 'descifrar')
print(f"Descifrado: {descifrado}")
```

```python
from cryptography.fernet import Fernet
#Cifrado Fernet

key = Fernet.generate_key()
cifrar = Fernet(key)


mensaje = "La clave es Ra9"


mensaje_cifrado = cifrar.encrypt(mensaje.encode()) #Convertimos el mensaje a formato Byte
print("Mensaje Cifrado:", mensaje_cifrado)


mensaje_descifrado = cifrar.decrypt(mensaje_cifrado).decode() #Convertir el resultado a texto
print("Descifrado:", mensaje_descifrado)




```

```python
from Crypto.Cipher import AES
from Crypto.Util.Padding import pad, unpad
from Crypto.Random import get_random_bytes
import binascii

#Cifrado AES(EBC y CBC)

# Generar una llave de 16 bytes (128 bits)
llave = get_random_bytes(16)
mensaje = "Este mensaje no debe ser leido."
mensaje_bytes = mensaje.encode('utf-8')

# MODO ECB
cifrador_ecb = AES.new(llave, AES.MODE_ECB)
cifrado_ecb = cifrador_ecb.encrypt(pad(mensaje_bytes, 16))
print("=== MODO ECB ===")
print(f"Cifrado (hex): {binascii.hexlify(cifrado_ecb).decode()}")

# Descifrar ECB
descifrador_ecb = AES.new(llave, AES.MODE_ECB)
descifrado_ecb = unpad(descifrador_ecb.decrypt(cifrado_ecb), 16)
print(f"Descifrado: {descifrado_ecb.decode()}")

# MODO CBC
iv = get_random_bytes(16)  # Vector de inicialización
cifrador_cbc = AES.new(llave, AES.MODE_CBC, iv=iv)
cifrado_cbc = cifrador_cbc.encrypt(pad(mensaje_bytes, 16))
print("\n=== MODO CBC ===")
print(f"Cifrado (hex): {binascii.hexlify(cifrado_cbc).decode()}")

# Descifrar CBC
descifrador_cbc = AES.new(llave, AES.MODE_CBC, iv=iv)
descifrado_cbc = unpad(descifrador_cbc.decrypt(cifrado_cbc), 16)
print(f"Descifrado: {descifrado_cbc.decode()}")

```

```python
from hashlib import sha256
from datetime import datetime

#Blockchain
class Block:
    def __init__(self, previous_hash_block, data):
        self.timestamp = str(datetime.now())
        self.previous_hash_block = previous_hash_block
        self.data = data

    def hash(self):
        info = "#".join(self.data) + self.timestamp + self.previous_hash_block
        return sha256(info.encode()).hexdigest()


class BlockChain:
    def __init__(self):
        self.chain = []

    def add(self, block):
        self.chain.append(block)

    def es_valida(self):
        for i in range(1, len(self.chain)):
            actual = self.chain[i]
            anterior = self.chain[i - 1]
            if actual.previous_hash_block != anterior.hash():
                print(f"La cadena se rompió entre el bloque {i-1} y el bloque {i}.")
                return False
        print("La cadena es válida.")
        return True


transacciones1 = ["Juan Diego le envia 5 mil a Luis", "Luis le envia 3 mil a Joaquin", "Luis le envia 2 mil a Juanes"]
transacciones2 = ["Juan Diego le envia 2 mil a German", "Luis le envia 2 mil a Tomas", "Luis le envia 2 mil a Martha"]
transacciones3 = ["Joaquin le envia 1 mil a Martha"]

block_chain = BlockChain()

b1 = Block("0" * 64, transacciones1)  # el primer bloque no tiene anterior real, usamos 64 ceros
block_chain.add(b1)

b2 = Block(b1.hash(), transacciones2)
block_chain.add(b2)

b3 = Block(b2.hash(), transacciones3)
block_chain.add(b3)

for i, b in enumerate(block_chain.chain): #Muestra los Hashes de los bloques
    print(f"Bloque {i}: Hash = {b.hash()}")

for i in range(1, len(block_chain.chain)):
    actual = block_chain.chain[i]
    anterior = block_chain.chain[i - 1]
    print(f"Bloque {i}: ¿el anterior hash coincide con el bloque {i-1}? "
          f"{actual.previous_hash_block == anterior.hash()}") #Esta linea de codigo mientras sea Verdadera verifica que la cadena este intacta



print("\nValidando cadena original:")
block_chain.es_valida()  # True

#Romper la cadena
print("\nHash de b2 ANTES de modificar:", b2.hash())
b2.data = ["Luis le envia 2 MILLONES a Tomas"]  # alteramos los datos del bloque 2
print("Hash de b2 DESPUÉS de modificar:", b2.hash())
print("¿b3 sigue apuntando al hash viejo de b2?", b3.previous_hash_block == b2.hash())


print("\nValidando cadena modificada:")
block_chain.es_valida()
```

```python
import requests
import ssl
import socket


url_https = "https://www.python.org"
url_http = "http://httpforever.com"

respuesta_https = None #Variables donde guardaremos los datos de las paginas
respuesta_http = None

print("Petición HTTPS ")
try:
    respuesta_https = requests.get(url_https, timeout=5)
    print(f"URL final: {respuesta_https.url}")  #Pide la URL de la pagina
    print(f"Código de estado: {respuesta_https.status_code}") #Su codigo de estado o codigo de respuesta
    print(f"¿Usa SSL/TLS?: {respuesta_https.url.startswith('https')}") #Verificación de SSL/TTS
except requests.exceptions.RequestException as e:
    print(f"Error en la petición HTTPS: {e}")

print("\n Petición HTTP ")
try:
    respuesta_http = requests.get(url_http, timeout=5)
    print(f"URL final: {respuesta_http.url}")
    print(f"Código de estado: {respuesta_http.status_code}")
    print(f"¿Usa SSL/TLS?: {respuesta_http.url.startswith('https')}")
except requests.exceptions.RequestException as e:
    print(f"Error en la petición HTTP: {e}")



print("\n Encabezados de la respuesta HTTPS ")
if respuesta_https is not None: #veririficar si la petición funciono
    for header, valor in respuesta_https.headers.items():
        print(f"{header}: {valor}")
else:
    print("No se pudo obtener respuesta HTTPS.")

print("\n Encabezados de la respuesta HTTP ")
if respuesta_http is not None:
    for header, valor in respuesta_http.headers.items():
        print(f"{header}: {valor}")
else:
    print("No se pudo obtener respuesta HTTP")


#Certificado
def revisar_certificado(dominio, puerto=443):
    """Se conecta al dominio e imprime los datos del certificado SSL/TLS."""
    contexto = ssl.create_default_context()
    with socket.create_connection((dominio, puerto), timeout=5) as sock:
        with contexto.wrap_socket(sock, server_hostname=dominio) as ssock: #abre las conexiones y las cierra automaticamente
            cert = ssock.getpeercert()
            print(f"\n=== Certificado de {dominio} ===")
            print(f"Emitido para: {cert.get('subject')}") #a quien pertenece el certificado
            print(f"Emitido por: {cert.get('issuer')}") #que entidad certificadora lo firmo
            print(f"Válido desde: {cert.get('notBefore')}") #Rango de fechas
            print(f"Válido hasta: {cert.get('notAfter')}")
            print(f"Versión TLS usada: {ssock.version()}") #Versión de TLS es usada


revisar_certificado("www.python.org") #Llama la función de mas arriba




```
